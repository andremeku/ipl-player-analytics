# IPL player analytics — guided exploration

Reproducible notebook based on user-supplied historical IPL datasets. Run from the repository root, or adjust the data directory below. Original project: team coursework; Andre contributed visualizations and conclusions.

````python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA = Path("data") if Path("data").exists() else Path("../data")
print("Data directory:", DATA.resolve())
````

## 1. Load supplied source files

````python
deliveries = pd.read_csv(DATA / "deliveries.csv")
players = pd.read_excel(DATA / "Players.xlsx")
consistency = pd.read_csv(DATA / "Player_Consistency_Stats.csv")
clusters = pd.read_csv(DATA / "Players_with_clusters.csv")
home_away = pd.read_csv(DATA / "teamwise_home_and_away.csv")
print("Delivery records:", len(deliveries))
print("Players:", len(players))
deliveries.head()
````

## 2. Top historical batters by runs

These values describe the supplied dataset only, not the current IPL season.

````python
top10 = deliveries.groupby("batsman")["batsman_runs"].sum().nlargest(10).sort_values()
ax = top10.plot.barh(figsize=(10, 5), title="Total runs in available delivery records")
ax.set_xlabel("Runs")
plt.tight_layout(); plt.show()
````

## 3. Demographics and batting statistics

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
players["Country"].fillna("Not recorded").value_counts().head(7).plot.bar(ax=axes[0], title="Player country")
consistency.query("total_matches >= 10").plot.scatter(x="strike_rate", y="average_runs", ax=axes[1], alpha=0.4, title="Batting metrics")
plt.tight_layout(); plt.show()
````

## 4. Supplied player clusters and team comparisons

````python
display(clusters["cluster"].value_counts(dropna=False).sort_index().rename("Players"))
display(home_away[["team", "home_win_percentage", "away_win_percentage"]].head(10))
````
