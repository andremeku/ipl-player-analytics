```python
# CORRELATION ANALYSIS
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# FILE PATHS
file_paths = {
    "players": r"C:\Users\kotha\Downloads\archive (3)\Players.xlsx",
    "deliveries": r"C:\Users\kotha\Downloads\archive (3)\deliveries.csv",
    "matches": r"C:\Users\kotha\Downloads\archive (3)\matches.csv",
    "most_runs": r"C:\Users\kotha\Downloads\archive (3)\most_runs_average_strikerate.csv",
    "teams": r"C:\Users\kotha\Downloads\archive (3)\teams.csv",
    "teamwise_home_and_away": r"C:\Users\kotha\Downloads\archive (3)\teamwise_home_and_away.csv",
}

# LOAD DATASETS
players_df = pd.read_excel(file_paths["players"])
deliveries_df = pd.read_csv(file_paths["deliveries"])
matches_df = pd.read_csv(file_paths["matches"])
most_runs_df = pd.read_csv(file_paths["most_runs"])
teams_df = pd.read_csv(file_paths["teams"])
teamwise_home_away_df = pd.read_csv(file_paths["teamwise_home_and_away"])

# FUNCTION TO COMPUTE AND PLOT CORRELATION
def plot_correlation_heatmap(dataframe, title):
    corr_matrix = dataframe.corr()
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, square=True)
    plt.title(title)
    plt.show()

# FILTER DATASETS FOR NUMERICAL FEATURES
numerical_dfs = {
    "Deliveries": deliveries_df.select_dtypes(include=['int64', 'float64']),
    "Matches": matches_df.select_dtypes(include=['int64', 'float64']),
    "Most Runs": most_runs_df.select_dtypes(include=['int64', 'float64']),
    "Teamwise Home and Away": teamwise_home_away_df.select_dtypes(include=['int64', 'float64']),
}

# PLOT CORRELATION FOR EACH DATASET
for name, df in numerical_dfs.items():
    if not df.empty:  # Ensure the dataframe contains numerical data
        plot_correlation_heatmap(df, f"Correlation Heatmap: {name}")
```
